# 🔑 Parte B — Solución comentada

### ✅ Paso 1 — Variables y tipos de datos

```python
aluna = "Ana"
nota = 8.5
modulos = 3
aprovada = nota >= 7

print(type(aluna), type(nota), type(modulos), type(aprovada))
print(f"{aluna} sacó {nota} → ¿aprobada? {aprovada}")
```

### ✅ Paso 2 — Listas

```python
notas_ana = [8.5, 6.0, 9.0]
print(len(notas_ana), sum(notas_ana), max(notas_ana), min(notas_ana))
print(notas_ana[0], notas_ana[-1], notas_ana[0:2])

notas_ana.append(7.5)
media = sum(notas_ana) / len(notas_ana)
print(f"Notas: {notas_ana} → media {media:.2f}")
```

### ✅ Paso 3 — Diccionarios

```python
turma = {
    "Ana": [8.5, 6.0, 9.0],
    "Bruno": [5.5, 7.0, 6.5],
    "Carla": [9.5, 9.0, 8.0],
    "Diego": [6.0, 5.0, 7.0],
    "Eduarda": [7.5, 8.5, 9.5],
    "Felipe": [4.0, 6.5, 5.0],
    "Gabriela": [8.0, 7.5, 7.0],
    "Henrique": [9.0, 4.5, 8.0],
}
print(turma["Carla"])
print(list(turma.keys()))

turma["Isabela"] = [10.0, 9.5, 9.0]
print(f"Ahora la turma tiene {len(turma)} alumnos")
```

### ✅ Paso 4 — Bucles (for) y condiciones (if)

```python
for aluno, notas in turma.items():
    media = sum(notas) / len(notas)
    if media >= 7:
        situacao = "Aprobado ✅"
    elif media >= 5:
        situacao = "Recuperación ⚠️"
    else:
        situacao = "Reprobado ❌"
    print(f"{aluno:<10} media {media:.2f} → {situacao}")
```

### ✅ Paso 5 — Funciones

```python
def calcular_media(notas):
    """Devuelve el promedio de una lista de notas."""
    return sum(notas) / len(notas)

def situacao(media, corte=7):
    """Clasifica la media del alumno."""
    if media >= corte:
        return "Aprobado"
    if media >= 5:
        return "Recuperación"
    return "Reprobado"

print(calcular_media([7, 8, 9]), situacao(6.2), situacao(6.2, corte=6))
```

### ✅ Paso 6 — Comprensiones de lista y diccionario

```python
medias = {aluno: round(calcular_media(n), 2) for aluno, n in turma.items()}
aprobados = [aluno for aluno, m in medias.items() if m >= 7]
ranking = sorted(medias.items(), key=lambda par: par[1], reverse=True)

print("Medias:", medias)
print("Aprobados:", aprobados)
print("Ranking:", ranking[:3])
```

### ✅ Paso 7 — Lo mismo, pero con pandas

```python
import pandas as pd

df = pd.DataFrame(turma, index=["Estatística", "Python", "BI"]).T
df["media"] = df[["Estatística", "Python", "BI"]].mean(axis=1).round(2)
df["situacao"] = df["media"].apply(situacao)
df = df.sort_values("media", ascending=False)
display(df)

print("Media de cada módulo:")
print(df[["Estatística", "Python", "BI"]].mean().round(2))
print("Menos de 6 en Python:", df[df["Python"] < 6].index.tolist())
```

### ✅ Paso 8 — Tu primer gráfico

```python
import plotly.express as px

fig = px.bar(df.reset_index(names="aluno"), x="aluno", y="media",
             title="Media final por alumno", labels={"aluno": "", "media": "Media"})
fig.update_traces(marker_color="#2a78d6", marker_cornerradius=4)
fig.add_hline(y=7, line_dash="dot", line_color="#52514e",
              annotation_text="Nota mínima 7,0", annotation_position="top left")
fig.update_layout(template="plotly_white", yaxis_range=[0, 10])
fig.show()
```

### ✅ Paso 9 — Guardar y leer un archivo CSV

```python
df.to_csv("notas_turma.csv")
de_novo = pd.read_csv("notas_turma.csv", index_col=0)
de_novo.head()
```
