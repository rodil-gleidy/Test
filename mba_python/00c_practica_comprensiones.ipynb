## Una comprensión es un `for` que construye algo

Este `for` crea una lista nueva, elemento por elemento:

```python
numeros = [1, 2, 3]
dobles = []
for n in numeros:
    dobles.append(n * 2)
```

La comprensión hace **lo mismo en una línea**:

```python
dobles = [n * 2 for n in numeros]
```

## Cómo traducir un `for` a comprensión

1. Abre **corchetes** `[ ]`. Reemplazan al `dobles = []` y al `.append`.
2. Lo que iba dentro de `.append(...)` va **primero**: `n * 2`.
3. Después copias la línea del `for` **tal cual, sin los dos puntos**: `for n in numeros`.
4. Si había un `if`, va **al final**, también sin dos puntos.

## Los moldes

| quiero crear… | escribo |
|---|---|
| una lista | `[QUÉ_GUARDO for NOMBRE in LO_QUE_RECORRES]` |
| una lista filtrada | `[QUÉ_GUARDO for NOMBRE in LO_QUE_RECORRES if CONDICIÓN]` |
| un diccionario | `{LLAVE: VALOR for NOMBRE in LO_QUE_RECORRES}` |
| un diccionario filtrado | `{LLAVE: VALOR for NOMBRE in LO_QUE_RECORRES if CONDICIÓN}` |

La parte `for NOMBRE in LO_QUE_RECORRES` es **idéntica** al `for` que ya conoces: entre `for` e `in` solo van nombres, y con un diccionario usas `.items()` para tener la llave y el valor.

## ¿Comprensión o `for` normal?

- **Comprensión:** cuando el resultado es una **lista o un diccionario nuevo** que vas a guardar en una variable.
- **`for` normal:** cuando quieres **imprimir** línea por línea o hacer varias cosas en cada vuelta.
- Nunca pongas un `print` dentro de una comprensión: para eso existe el `for` normal.

## Errores típicos

| escribiste | Python dice | corrección |
|---|---|---|
| `[n for n * 2 in numeros]` | `cannot assign to expression` | la cuenta va primero: `[n * 2 for n in numeros]` |
| `[n * 2 for n in numeros:]` | `invalid syntax` | sin dos puntos dentro de la comprensión |
| `{nombre, len(nombre) for nombre in nombres}` | `did you forget parentheses…` | llave y valor se separan con `:` y no con coma |

> Al imprimir una lista de textos, Python muestra comillas simples: `['Ana', 'Bruno']`. Es normal, no es un error.
