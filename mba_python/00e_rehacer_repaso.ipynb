### ✅ Ejercicio 1 — 🔍 Detective: el diccionario que desaparece

````python
gastos = {"enero": [100, 50], "febrero": [80, 20, 30]}
for mes, valores in gastos.items():
    print(mes, sum(valores))
print(gastos)
````

**Error 3.** El `for` usó `gastos` como nombre para la lista de cada mes, así que en cada vuelta **pisó** el diccionario. Al terminar, `gastos` quedó con la última lista. Los nombres que inventas entre `for` e `in` tienen que ser **distintos** de los que ya usas.

### ✅ Ejercicio 2 — ✍️ Desde cero: precios en dólares

````python
productos = {"café": 12.0, "pan": 8.0, "queso": 35.0, "leche": 6.5}
en_dolares = {nombre: round(precio / 5, 2) for nombre, precio in productos.items()}
print(en_dolares)
````

**Error 4.** Llaves **con** `nombre: valor` dan un diccionario. Llaves **sin** `:` (solo el valor) dan un conjunto, sin los nombres. Y entre `for` e `in` va un nombre **nuevo** (`nombre`), no `productos`, que ya es el diccionario (**error 3**).

### ✅ Ejercicio 3 — ✍️ Desde cero: función + comprensión

````python
totales = {"Ana": 250, "Bruno": 120, "Carla": 200}

def clasificar(valor):
    if valor >= 200:
        return "grande"
    return "chico"

tipos = {nombre: clasificar(valor) for nombre, valor in totales.items()}
print(tipos)
````

- El nombre entre los paréntesis del `def` es la "caja" donde llega el número cuando llamas a la función. Sin él, `valor` no existe dentro de la función (**error 3**).
- "200 **o más**" es `>=`: con `>` Carla quedaría como "chico".
- Sin `.items()`, recorrer el diccionario solo da los nombres. Con `.items()` tienes los dos.
- Los nombres no tienen que coincidir: si en el `for` usas `num`, `clasificar(num)` mete ese número en la caja `valor`.

### ✅ Ejercicio 4 — ✍️ Desde cero: lo mismo en pandas

````python
import pandas as pd
compras = pd.DataFrame({"cliente": ["Ana", "Bruno", "Carla"], "valor": [250, 120, 200]})
compras["tipo"] = compras["valor"].apply(clasificar)
grandes = compras[compras["tipo"] == "grande"]
print(grandes)
````

Tres reglas en tres líneas: **error 1** (guardar la columna con `compras["tipo"] =` y el filtro con `grandes =`), **error 2** (`"grande"` con comillas porque es un texto fijo) y `clasificar` **sin paréntesis** dentro de `.apply()`, porque se la pasas a pandas para que la ejecute él.
