## La receta del `for`

Antes de escribir cualquier `for`, respóndete tres preguntas:

1. **¿Qué voy a recorrer?** Una lista, un diccionario, un `range(...)`.
2. **¿Cómo llamo a cada elemento?** Un nombre que tú inventas: `fruta`, `nota`, `producto`.
3. **¿Qué hago con cada uno?** Imprimirlo, sumarlo, compararlo.

Y rellenas este molde:

```python
for  NOMBRE_QUE_INVENTAS  in  LO_QUE_RECORRES:
    LO_QUE_HACES_CON_CADA_UNO
```

- La línea del `for` **termina en dos puntos** `:`.
- Lo que se repite va **indentado** (4 espacios).
- El nombre que pones después de `for` **lo crea el `for`**: no hace falta haberlo definido antes.

## Las tres formas que vas a usar

| quiero recorrer… | escribo | en cada vuelta tengo |
|---|---|---|
| una lista | `for nota in notas:` | un elemento |
| las llaves de un diccionario | `for producto in precios:` | una llave |
| los pares de un diccionario | `for producto, precio in precios.items():` | la llave y su valor |

### ✅ Ejercicio 14 — Desafío: alerta de stock

````python
estoque = {"café": 0, "pan": 15, "leche": 3, "queso": 8, "mantequilla": 2}
reponer = 0
for producto, cantidad in estoque.items():
    if cantidad < 5:
        print(f"⚠️ Reponer {producto} ({cantidad} unidades)")
        reponer = reponer + 1
print(f"Productos para reponer: {reponer}")
````
