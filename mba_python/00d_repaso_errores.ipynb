## Los 4 errores

| error | cómo se ve | regla |
|---|---|---|
| **1. Calcular sin guardar** | `palabra.upper()` suelto, `df.mean(...)` sin `df["media"] =` | Si no hay un `=` a la izquierda, el resultado se pierde |
| **2. Comillas de más** | `letras["nombre1"]` en vez de `letras[nombre1]` | Con comillas es un texto fijo; sin comillas es la variable |
| **3. Lo que existe vs. lo que inventas** | `for t in temperatura` (no existe), `for mes, gastos in gastos.items()` | Después del `in`: algo que ya existe. Entre `for` e `in`: nombres nuevos, distintos de los que ya usas |
| **4. Corchetes vs. llaves** | una lista cuando querías un diccionario, `{x for ...}` | `[ ]` crea una lista · `{llave: valor}` un diccionario · `{x}` sin dos puntos, un conjunto |

## Del mensaje de error al tipo de error

| ves esto | probablemente es |
|---|---|
| El resultado no cambió y no hubo error | **1**: calculaste pero no guardaste |
| `AttributeError: 'NoneType' object has no attribute 'append'` | **1** al revés: escribiste `lista = lista.append(...)` |
| Un diccionario con una sola llave rara, como `{'producto': 0}` | **2**: comillas en la variable |
| `KeyError: 'nombre'` | **2**: buscaste el texto "nombre" en vez de la variable |
| `NameError: name '...' is not defined` | **3**: el nombre no existe, o está mal escrito (o no ejecutaste la celda que lo crea) |
| `SyntaxError` dentro de una comprensión | **4**: dos puntos dentro de `[ ]`, o coma en vez de dos puntos dentro de `{ }` |

### ✅ Ejercicio 10 — 🔍 Detective: ¿qué crea cada línea?

````python
numeros = [1, 2, 3]
a = [n * 2 for n in numeros]       # a es: lista        → [2, 4, 6]
b = {n: n * 2 for n in numeros}    # b es: diccionario  → {1: 2, 2: 4, 3: 6}
c = {n * 2 for n in numeros}       # c es: conjunto     → {2, 4, 6}
print(type(a))
print(type(b))
print(type(c))
````

**Error 4.** Lo que decide qué se crea es el símbolo de afuera **y** si hay dos puntos adentro: `[ ]` → lista, `{llave: valor}` → diccionario, `{x}` sin dos puntos → conjunto.

### ✅ Ejercicio 14 — ✍️ Desde cero: lo mismo en pandas

````python
import pandas as pd
compras = pd.DataFrame({"cliente": ["Ana", "Bruno", "Carla"], "valor": [250, 120, 200]})
compras["tipo"] = compras["valor"].apply(clasificar)
grandes = compras[compras["tipo"] == "grande"]
print(grandes)
````

Tres reglas en tres líneas: **error 1** (guardar la columna con `compras["tipo"] =` y el filtro con `grandes =`), **error 2** (`"grande"` con comillas porque es un texto fijo) y `clasificar` **sin paréntesis** dentro de `.apply()`, porque se la pasas a pandas para que la ejecute él.

### ✅ Ejercicio 15 — 🔍 Desafío final: tres errores en dos líneas

- **Error 4:** `[nombre: ...]` usa corchetes con `llave: valor` → `SyntaxError`. Un diccionario va con llaves `{ }`.
